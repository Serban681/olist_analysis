```python
import pandas as pd

df = pd.read_csv('../runs/03_category_revenue.csv')

top_5_categories_df = df.iloc[0:5][['category', 'total_revenue']]

top_5_categories_df
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(
    figsize=(12,6)
)

ax.grid(axis="y", alpha=0.3)
ax.set_axisbelow(True)

ax.bar(
    top_5_categories_df['category'],
    top_5_categories_df['total_revenue'],
    color="#8C45A0",
    width=0.3
)

ax.ticklabel_format(style="plain", axis="y")

plt.tight_layout()
plt.xlabel('category')
plt.ylabel('total revenue')
plt.show()
```
