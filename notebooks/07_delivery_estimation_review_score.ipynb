```python
import pandas as pd

dlv_status_low_rvw_rate_df = pd.read_csv('../runs/07_delivery_estimation_review_score_3.csv')

dlv_status_low_rvw_rate_df.head()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(
    figsize=(5,3)
)

ax.grid(axis="y", alpha=0.3)
ax.set_axisbelow(True)

ax.bar(
    dlv_status_low_rvw_rate_df['delivery_status'],
    dlv_status_low_rvw_rate_df['low_review_rate'],
    color="#B72438",
    width=0.3
)

ax.ticklabel_format(style="plain", axis="y")

plt.tight_layout()
plt.xlabel('status')
plt.ylabel('low review rate')
plt.show()
```
